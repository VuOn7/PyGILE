## 1. `while` loops: repeat until something changes

A `for` loop runs once for each item in a list. You know in advance how many times it will run.

Sometimes you do not know that. You want to repeat **until** something happens. A `while` loop does this:

```
while condition:
    body
```

Python checks the condition. If it is `True`, the body runs, and Python checks again. When the condition becomes `False`, the loop stops.

Something in the body must change the condition. If nothing does, the loop never ends: an **infinite loop**. In JupyterLab, stop it with the ■ button (Kernel → Interrupt).

**Predict:** a glacier of 12 km² loses 2 % of its area every year (made-up values for practice). After how many years is it smaller than half its size? Write down a guess, then run the cell.

````python
area_km2 = 12.0        # made-up glacier area
loss_rate = 0.02       # loses 2 % of its area each year (made-up)
years = 0

while area_km2 >= 6.0:                     # still at least half the start size?
    area_km2 = area_km2 * (1 - loss_rate)  # one more year of melting
    years += 1

print(years, "years; area now", round(area_km2, 2), "km²")
````

````python
months = ["jan", "feb", "mar", "apr", "may", "jun", "jul", "aug", "sep", "oct", "nov", "dec"]
station_a_mm = [35, 59, 78, 151, 206, 536, 1045, 952, 592, 198, 42, 42]   # practice_rainfall.csv (made-up)
station_b_mm = [16, 23, 28, 54, 79, 246, 384, 394, 222, 79, 14, 14]

total_mm = 0
for i in range(12):
    total_mm += station_a_mm[i]
    if total_mm > 1000:
        print(f"Station A passes 1,000 mm in {months[i]}: {total_mm} mm so far")
        break                              # stop the loop now
````

The glacier needs 35 years to fall below 6 km². Many people guess 25, because 2 % of 12 km² is 0.24 km² and 6 ÷ 0.24 = 25. But each year loses 2 % of a **smaller** area, so the loss slows down.

The second cell shows `break`: it ends a loop at once, even a `for` loop. Station A passes 1,000 mm in June.

**Which loop?** Use `for` when you go through a list or a known number of steps. Use `while` when you stop on a condition and cannot say in advance how many steps that takes.

## 2. Tuples, unpacking and `zip`

You met **tuples** briefly in A8: values in round brackets, such as `(27.9881, 86.9250)`. A tuple is like a list that cannot be changed after it is made. That makes it a good home for values that belong together, such as a coordinate pair.

**Unpacking** puts the values of a tuple into separate names in one line: `lat, lon = everest`. The number of names must match the number of values.

A function can return two values by writing `return a, b`. Python packs them into a tuple, and you can unpack them straight away.

`zip(list_1, list_2)` walks through two lists side by side and gives one pair per step.

**Write the order down.** Here the pair is (lat, lon). Many map tools want (lon, lat) instead, and mixing them up is one of the most common GIS mistakes (Part B).

````python
everest = (27.9881, 86.9250)      # (lat, lon) from nepal_peaks.csv
print(type(everest), everest[0])

lat, lon = everest                # unpacking
print(f"Everest: lat {lat}, lon {lon}")
````

````python
def rain_range(rain_mm):
    """Return the smallest and the largest monthly rainfall."""
    return min(rain_mm), max(rain_mm)

low_mm, high_mm = rain_range(station_a_mm)
print(low_mm, high_mm)

places = ["Lukla", "Namche Bazaar", "Everest Base Camp"]
elevations_m = [2860, 3440, 5364]
for place, elevation_m in zip(places, elevations_m):
    print(f"{place}: {elevation_m:,} m")
````

`everest` is a `tuple`, and indexing works as with lists. `rain_range` returned two values, which landed in `low_mm` (35) and `high_mm` (1,045). `zip` gave the loop one (place, elevation) pair at a time.

## 3. Sets and dictionary comprehensions

A **set** is a collection with **no duplicates** and no order. Write it with curly brackets, `{"Dolakha", "Rasuwa"}`, or make one from a list with `set(...)`. Every repeated value is kept once.

Sets are made for comparing groups of names:

- `a - b` (**difference**): in `a` but not in `b`,
- `a & b` (**intersection**): in both,
- `a | b` (**union**): in either.

A **dictionary comprehension** builds a dictionary in one line, like the list comprehensions of A3: `{key: value for ... in ...}`. You can add `if` at the end to keep only some items.

**Predict:** `practice_landslides.csv` has 40 records. How many **different** district names do you expect?

````python
import pandas as pd
from pathlib import Path

DATA = Path("..") / "data"
slides = pd.read_csv(DATA / "practice_landslides.csv")    # made-up practice records

landslide_names = set(slides["district"])
print(len(slides), "records,", len(landslide_names), "districts")
print(sorted(landslide_names))

# district names as the district map of Part B spells them (typed here by hand)
map_names = {"Chitawan", "Dolakha", "Gorkha", "Kaski", "Kathmandu", "Myagdi", "Rasuwa", "Sindhupalchok", "Taplejung"}
print(landslide_names - map_names)      # in the landslide file, missing from the map
print(map_names - landslide_names)      # on the map, but no landslide records
````

````python
heights_m = {"Everest": 8849, "Kanchenjunga": 8586, "Lhotse": 8516, "Makalu": 8485}

heights_ft = {name: round(h_m * 3.28084) for name, h_m in heights_m.items()}
above_8500 = {name: h_m for name, h_m in heights_m.items() if h_m > 8500}
print(heights_ft)
print(above_8500)
````

The 40 records hold only 8 different districts. The set difference finds a real problem at once: the landslide file writes **Chitwan**, but the map writes **Chitawan**. A join between them would lose those 9 records. You will fix this in B6.

The second difference shows the map names with no records: Chitawan (because of the spelling) and Kathmandu.

The dictionary comprehensions made a table of heights in feet, and a smaller one with only the three peaks above 8,500 m.

## 4. Your own module

In A5 you wrote useful functions, but they live inside one notebook. To use them in the next notebook, you would copy them, and soon you have five slightly different copies.

A **module** is a `.py` file of Python code, usually functions. You write it once and `import` it wherever you need it, exactly like `import math`.

Normally you would create the file in JupyterLab with File → New → Python File. Here the code writes it for you with `write_text` (A7), so the lesson runs from top to bottom. The file goes into `../outputs`.

Python looks for modules in a list of folders called **`sys.path`**. The notebook's own folder is on that list, but `../outputs` is not, so you add it once with `sys.path.insert(0, ...)`. The `0` puts it first in the list.

````python
OUT = Path("..") / "outputs"
OUT.mkdir(exist_ok=True)

# The whole module is one string. Its outer quotes are ''' so that the
# docstrings inside it can use the usual triple double quotes.
module_text = '''"""Small tools for slopes, heights and treks (PyGILE lesson A12)."""
import math


def slope_deg(rise_m, run_m):
    """Return the slope angle in degrees from a rise and a run in metres."""
    return math.degrees(math.atan(rise_m / run_m))


def m_to_ft(length_m):
    """Return a length in metres converted to feet."""
    return length_m * 3.28084


def days_needed(climb_m, gain_per_day_m=500):
    """Return the days a climb takes; 500 m per day is a practice value."""
    return climb_m / gain_per_day_m
'''

tools_file = OUT / "nepal_tools.py"
tools_file.write_text(module_text, encoding="utf-8")
print(tools_file, tools_file.exists())
````

````python
import sys
sys.path.insert(0, str(OUT))    # also look for modules in ../outputs

import nepal_tools

print(round(nepal_tools.slope_deg(300, 500), 1))
print(round(nepal_tools.m_to_ft(8849)))
help(nepal_tools.days_needed)
````

`import nepal_tools` read the file and made its functions available with a dot, as with `math.tan`. A 300 m rise over 500 m is a 31.0° slope (A5), and Everest is 29,032 ft. `help` shows the docstring you wrote, so the docstring is your module's manual.

You may notice a new `__pycache__` folder in `outputs`. Python stores a translated copy of the module there to load it faster. You can ignore it.

**Editing a module.** Python reads a module only **once** per session. Below, a new function is added to the file. **Predict:** can the notebook use it straight away?

````python
module_text_v2 = module_text + '''

def ft_to_m(length_ft):
    """Return a length in feet converted to metres."""
    return length_ft / 3.28084
'''
tools_file.write_text(module_text_v2, encoding="utf-8")
print("ft_to_m" in module_text_v2)     # the new function is in the file text
````

````python
nepal_tools.ft_to_m(29032)
````

`AttributeError: module 'nepal_tools' has no attribute 'ft_to_m'`. The file on disk has the new function, but the notebook still holds the old copy it read at `import`. Running `import nepal_tools` again does nothing, because Python sees the module is already loaded.

`importlib.reload` reads the file again:

````python
import importlib

importlib.reload(nepal_tools)
print(round(nepal_tools.ft_to_m(29032)))
````

After the reload, `ft_to_m(29032)` gives back 8,849 m. Two other ways to import are common: `from nepal_tools import m_to_ft` brings in one function by name, and `import nepal_tools as nt` gives a short nickname, like `import pandas as pd`.

## 5. Running a script

A module is a file you **import** to borrow its functions. A **script** is a file you **run** to do a whole job, from the first line to the last. It is like a notebook with all cells joined into one, and no outputs saved inside.

Scripts are good for jobs you repeat: the same summary every month, or the same steps on twenty files.

In a notebook, `%run file.py` runs a script. Lines that start with `%` are **magic commands**: special commands of Jupyter, not Python. In a terminal (or the Anaconda Prompt on Windows), you would type `python rain_summary.py` in the folder that holds the file.

The script below uses a relative path, `../data/practice_rainfall.csv`. A relative path starts from the working directory (A7). That works here, and also in a terminal inside `outputs`, because both folders sit next to `data`.

````python
script_text = '''"""Print the annual rainfall of the practice stations (made-up data)."""
import pandas as pd

rain = pd.read_csv("../data/practice_rainfall.csv")
rain["annual_mm"] = rain.loc[:, "jan_mm":"dec_mm"].sum(axis=1)
ranked = rain.sort_values("annual_mm", ascending=False)

print("Annual rainfall, practice stations (made-up data)")
print(ranked[["station", "district", "annual_mm"]])
print("Wettest:", ranked.iloc[0]["station"])
'''

script_file = OUT / "rain_summary.py"
script_file.write_text(script_text, encoding="utf-8")
print(script_file.exists())
````

````python
%run ../outputs/rain_summary.py
````

The script read the table, added an annual total per station, and printed them from wettest to driest. Station A is the wettest, with 3,936 mm.

After `%run`, the script's variables (`rain`, `ranked`) also exist in the notebook. That is handy for checking, but remember where they came from.

## 6. Getting help on your own

Nobody remembers every function. Professionals look things up all day. These tools answer most questions:

- `help(thing)` prints the docstring of a function or module (A5).

## 7. Notebooks you can trust

A notebook remembers every variable you have created since the **kernel** started. The kernel is the Python process behind the notebook. This memory is called **hidden state**, because you cannot see it on the page.

Hidden state causes strange results. You delete a cell, but its variable lives on. You run cells out of order, and a number changes. The notebook works for you today and fails for a colleague tomorrow.

The cure is a habit: before you trust or share a result, use **Kernel → Restart Kernel and Run All Cells**. The numbers in the brackets on the left, such as `[7]`, show the order in which cells ran; after a clean run they count up from 1.

`%whos` lists every variable the kernel holds right now.

**Try it:** run the next two cells. Then delete the first one (click it, then Edit → Delete Cells) and run the second again.

````python
secret_rain_mm = 250    # made-up value: delete this cell after running it
````

````python
print(secret_rain_mm * 2)
````

````python
%whos int
````

The second cell still prints 500 after you delete the first one: `secret_rain_mm` stays in the kernel's memory. `%whos int` lists it among the whole numbers. After **Restart and Run All**, the second cell would stop with `NameError`, and only then would you see the problem.

**Random numbers.** Part D uses random samples. A computer's "random" numbers come from a formula that starts at a **seed**. The same seed gives the same numbers on every run and every computer, so others can repeat your results. Such results are called **reproducible**.

````python
import numpy as np

rng = np.random.default_rng(42)            # a random number generator with seed 42
print(rng.integers(0, 100, size=5))        # five whole numbers from 0 to 99

rng = np.random.default_rng(42)            # the same seed again
print(rng.integers(0, 100, size=5))        # the same five numbers
````

Both lines print `[ 8 77 65 43 43]`. Without a seed, `np.random.default_rng()` would give different numbers on every run. Always set a seed when your results depend on random numbers.

## 8. Checking code, yours and other people's

Code from a website, a colleague or an **AI assistant** can look perfect and still be wrong. So can your own. A wrong number in a report is worse than no number. Check code in four moves:

1. **Read it** line by line. Do you know what every line does?
2. **Predict** what it should give for a case you can work out by hand.
3. **Run it on that tiny case** and compare.
4. **Keep the check** as an `assert` statement.

`assert condition, "message"` does nothing when the condition is `True`. When it is `False`, it stops with an `AssertionError` and your message. A set of asserts is a small **test**: it checks your code each time you run the notebook.

The function below was suggested by an AI assistant for converting a gradient in percent to a slope angle. **Predict:** what should it give for 100 %? (A5 tells you.)

````python
import math

def percent_to_degrees(slope_percent):
    """Return a gradient in percent as a slope angle in degrees."""
    return math.atan(slope_percent / 100)

print(percent_to_degrees(100))
````

````python
assert round(percent_to_degrees(100), 6) == 45, "100 % should be 45 degrees"
````

The docstring promises degrees, and the code looks reasonable. But a 100 % gradient must be 45°, and the function gives 0.785. That is 45° in **radians**: the assistant forgot `math.degrees` (A5). The `assert` caught it, with the message you wrote.

The fixed version passes both checks silently:

````python
def percent_to_degrees(slope_percent):
    """Return a gradient in percent as a slope angle in degrees."""
    return math.degrees(math.atan(slope_percent / 100))

assert round(percent_to_degrees(100), 6) == 45, "100 % should be 45 degrees"
assert percent_to_degrees(0) == 0, "a flat road has no slope"
print("all checks passed")
````

Use AI assistants as a fast, confident colleague who is sometimes wrong. Ask them to explain each line, and never paste a number into a report that you have not checked.

**Keeping a history: version control.** When a project grows, you will want to know what changed, when, and why, and to go back to yesterday's working version. **Git** is the standard tool for this: it saves snapshots of a folder, called **commits**, each with a short message. **GitHub** is a website that stores Git projects online, so you can share them and work together; this course lives there. **GitHub Desktop** is a free program that lets you use Git with buttons instead of typed commands. ## 9. When it breaks: ModuleNotFoundError

Importing your own module goes wrong in a few typical ways. Here is the most common one. Run the cell.

````python
import nepal_tool
````

The last line reads `ModuleNotFoundError: No module named 'nepal_tool'`. Python searched every folder in `sys.path` and found no file called `nepal_tool.py`. Check three things, in this order:

1. **The spelling.** The file is `nepal_tools.py`, with an `s`. Module names are the file name without `.py`.
2. **The folder.** Is the file's folder in `sys.path`? A kernel restart forgets `sys.path.insert`, so run that cell again after every restart. `print(sys.path[0])` shows the first folder searched.
3. **The file.** `(OUT / "nepal_tools.py").exists()` should give `True`.

One more trap: never name your own file like a library, such as `math.py` or `pandas.py`. Python would import your file instead of the real library, and you would get confusing errors.

**Try it:** fix the name and run the cell again.

### Review 1 · Years to half, as a tested function

Write a function `years_to_half(loss_rate)` with a docstring. It uses a `while` loop to return the number of years until an area falls below half of its start (start with an area of 1.0). Test it with two asserts: a loss of 60 % a year (0.6) takes 1 year, and 2 % takes 35 years (step 1). Then build a dictionary for the rates 1 %, 2 %, 5 % and 10 % with a dictionary comprehension, and print it. The rates are made-up values for practice.

````python
# Your answer
````

<details>
<summary><b>Show solution</b></summary>

```python
def years_to_half(loss_rate):
    """Return the years until an area shrinking by loss_rate each year is below half."""
    area = 1.0
    years = 0
    while area >= 0.5:
        area = area * (1 - loss_rate)
        years += 1
    return years

assert years_to_half(0.6) == 1
assert years_to_half(0.02) == 35
print({rate: years_to_half(rate) for rate in [0.01, 0.02, 0.05, 0.10]})
```

`{0.01: 69, 0.02: 35, 0.05: 14, 0.1: 7}`. The function does not need the real area, because only the share that is left matters. Why 60 % and not 50 %? A 50 % loss leaves exactly half, which is not **below** half, so it takes 2 years: tiny tests also check the edges of a condition. The two asserts protect it: if you change the loop later and break it, the notebook stops with a clear message.
</details>

## Recap

- `while condition:` repeats until the condition is `False`; `break` leaves any loop early. Use `for` for lists, `while` for "until".
- Tuples hold fixed groups such as `(lat, lon)`; unpacking, `return a, b` and `zip` work with them. Sets remove duplicates and compare names with `-`, `&` and `|`.
- A **module** is a `.py` file of functions you `import`; add its folder to `sys.path`, and `importlib.reload` it after editing. A **script** is a file you run with `%run` or `python file.py`.
- Help is always near: `help()`, `?`, `dir()`, Tab, Shift+Tab, error messages and the official documentation.
- Trust comes from habits: Restart and Run All before sharing, a seed for random numbers, and `assert` checks on tiny cases you can work out by hand, for every piece of code you did not write and every piece you did.

### Words you learned

| Word | Meaning |
| --- | --- |
| `while` loop | A loop that repeats as long as a condition is `True` |
| infinite loop | A loop whose condition never becomes `False`; stop it with Kernel → Interrupt |
| `break` | Leaves the loop at once |
| tuple | An ordered group of values in round brackets that cannot be changed |
| unpacking | Putting the values of a tuple into separate names: `lat, lon = everest` |
| `zip` | Walks through two or more lists side by side, one tuple per step |
| set | A collection with no duplicates and no order, like `{"Dolakha", "Rasuwa"}` |
| difference / intersection / union | `a - b`, `a & b` and `a | b` on sets |
| dictionary comprehension | Builds a dictionary in one line: `{k: v for ... in ...}` |
| module | A `.py` file of code that you `import` |
| `sys.path` | The list of folders where Python looks for modules |
| `importlib.reload` | Reads an already imported module again, after you edit it |
| script | A `.py` file you run from top to bottom to do a job |
| magic command | A Jupyter command starting with `%`, such as `%run` and `%whos` |
| `dir()` | Lists the names inside a module or object |
| kernel | The Python process that runs a notebook's cells and holds its variables |
| hidden state | Variables in the kernel's memory that the page no longer shows |
| seed | The starting value of a random number generator; same seed, same numbers |
| reproducible | Gives the same result when someone else runs it |
| `assert` | Stops with an `AssertionError` if a condition is `False` |
| test | A small check that code gives the answer you know is right |
| Git / commit | A version control tool / one saved snapshot of a project with a message |
